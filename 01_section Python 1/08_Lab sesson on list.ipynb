```python
class Solution:
    def solve(self, A): #A=[3,1,2,4]
      # Step 1: Calculate the largest element
      # alternatively
     # max_ele=max(A)
      max_ele=-1 # -1
      for x in A: # x=3, 1, 2, 4
        if x>max_ele: # 3>-1 T, 1>3 F, 2>3 F, 4>3 T
          max_ele=x # max_ele=3 -> 4

      # Step 2: Calculate the second largest element
      sec_largest=float('-inf')
      for x in A: #3, 1, 2, 4
        if x>sec_largest and x!=max_ele: # 3>-1 T and 3!=4 T
          #1>3 F, 2>3 F, 4>3 and 4!=4
          sec_largest=x #  sec_largest=3

      return sec_largest
```
