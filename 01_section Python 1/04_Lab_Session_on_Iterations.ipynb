```python
# Check wheather the number is prime or not
- numbers divisible by 1 and itself - which mean total no of factors should be 2
- 1 is nethier prime nor composite
- if I want to check that a is factor or n -> n % a == 0 -> yes

2,3,5,7,11,13,17,19,23,...etc

Approach:
7 -> 1,2,3,4,5,6,7 possible factors
Now 1 and 7 are factors of 7
2,3,4,5,6 -> if this divides 7 completely then 7 is not a prime no.
No number in the range divides 7 -> 7 is Prime

9 -> 1,2,3,4,5,6,7,8,9 possible factors
Now 1 and 9 are factors of 9
2,3,4,5,6,7,8 -> if this divides 9 completely then 9 is not a prime no.
3 divides 9 -> 9 is not a Prime
```

```python
n=int(input())
Prime=True # Number is prime
for a in range(2,n):  # start=2, end=n-1
  if n%a==0:
    Prime=False # because a divides n, n is not prime
    break

if (Prime): #if True
  print("Number is prime")
else:
  print("Number is not prime")
```

```python
# Find the reverse of the given number

1356 -> 6531
154 -> 451
4378 -> 8734
-256 -> -652

- Get the last digit # d=n%10
- Add the last digit to the back of the reversed number # rev=rev*10+d
- Remove the last digit from the original number # n=n//10
- keep repeating till n>0


```

```python
n = int(input())
rev = 0
copy = n
if n<0:
    n=n*-1


while n>0:
    d = n%10
    rev = rev*10 +d
    n=n//10

if copy<0:
    rev=rev*-1
print(rev)
```
