# 2D list -> nested list which means a list inside a list

```python
# 2D list -> nested list which means a list inside a list
maths=[10,20,30] # row 1
science=[40,50,60] # row 2
history=[70,80,90] # row 3

subjects=[maths,science,history] # your whole table
print(subjects)  # [[10, 20, 30], [40, 50, 60], [70, 80, 90]]
```

```python
#indexing
#        0           1              2
# [[10, 20, 30], [20, 30, 40], [50, 60, 70]]
print(subjects[0])
print(subjects[1])
print(subjects[2])

#accesing each item
print(subjects[0][0]) #10
print(subjects[0][1]) #20
print(subjects[0][2]) #30
print(subjects[1][0]) #20
print(subjects[1][1]) #30
print(subjects[1][2]) #40
print(subjects[2][0]) #50
print(subjects[2][1]) #60
print(subjects[2][2]) #70
```

## iterating through 2D list


```python
# iterating over a 2 list - row wise printing
#subjects=[[1,1,1],[2,2,2],[3,3,3]]
# len(subjects[i]) -> len(subjects[0]) = 3
# len(subjects[1]) = 3
# len(subjects[2]) = 3

# len(subjects) -> no of rows in the list which is 3
for i in range(len(subjects)): # 0,1,2
  for j in range(len(subjects[i])): # i=0, j=0,1,2, i=1, j=0,1,2, i=3, j=0,1,2
    print(subjects[i][j],end=" ") #[0][0],[0][1],[0][2], [1][0],[1][1],[1][2]
    #[2][0],[2][1],[2][2]
  print()
```

```python
#Column wise printing
for i in range(len(subjects)):
  for j in range(len(subjects[i])):
    print(subjects[j][i],end=" ")
    #[2][0],[2][1],[2][2]
  print()
```

## Taking input for 2D list

```python
def take_list_as_input():
    # input() -> takes while string as value '1','2','3'
    # split() -> splits the string into list of strings ['1','2','3']
    # map(int,...) -> converts each string into integer [1,2,3]

    li = list(map(int, input().split()))
    return li

a=[] 
for i in range(3):
    a.append(take_list_as_input())

print(a)
```

```python
# Row wise sum : [[1, 2, 3], [4, 5, 6], [7, 8]]
# i=0, [1, 2, 3] -> sum([1, 2, 3])=6
# i=1, [4, 5, 6] -> sum([4, 5, 6])=15
# i=2, [7, 8] -> sum([7, 8])=15
for i in range(len(a)):
  row_sum= sum(a[i])
  print(row_sum)
```

```python
# find an element in the 2D list
# [[1,2,3],[4,5,6],[7,8,9]]
#k =5, True -> found
# K =10 Flase -> not found

k = int(input())
flag = False
for i in range(len(a)):
    for j in range(len(a[i])):
        if a[i][j] == k:
            print("Element Found")
            flag = True
            break
    
if not flag:
    print("Element Not Found")
```

```python
# approach 2 -> using in

k=int(input())
flag=False
for row in a: # row gives 1 1d list at a time
  if k in row:
    print("Element found")
    flag=True
    break

if not flag:
  print("Element not found")
```

```python
# find max value in an 1d list
# method 1
li=[-13,-53,-23,-21,-55]
max_value=li[0]
for item in li:
  if max_value<item:
    max_value=item

print(max_value)
```

```python
# method 2
li.sort() #sorts in ascending order
print(li[-1])
li.sort(reverse=True) #sorts in descending order
print(li[0])
```

```python
# method 3 - using the max function
print(max(li))
```

```python
# minimum value
print(min(li))
```

```python
# list comprehension

squares=[]
for i in range(1,6):
  squares.append(i**2)

print(squares)

# shorthand way of writing the above code
sq=[i**2 for i in range(1,6)]
print(sq)

#list_name=[value where condition]
```

```python
# even numbers from 1 to 10
evens = [i for i in range(1,11) if i%2==0]
print(evens)
```

```python
odds = [i for i in range(1,11) if i%2!=0]
print(odds)
```

```python
# pairs(i,j) where j<i
pairs=[(i,j)for i in range(3) for j in range(3) if j<i]
print(pairs)
```
