# Function

### Need of Function
A function is a reusable block of code that performs a specific task. It helps in organizing code, reducing redundancy, and improving readability. Functions allow you to break down complex problems into smaller, manageable pieces, making it easier to debug and maintain the code.

````python
print("*******************************")
print("Transcation Completed")
print("Thank you for visiting")
print("******************************")
````

### Syntax of Function
```python

    #action


#Call the function to use it
function_name()
```

````python
def message():
    print("*******************************")
    print("Transcation Completed")
    print("Thank you for visiting")
    print("******************************")

message()
print("new trans")

message()
print("new trans")

message()
print("Calling again")
````

### Return Statement

````python
def area_cicle(): # function definition
    print(3.14*5*5) 

area_cicle() #Calling the function
````

````python
def area_cicle(): # function definition
    return (3.14*5*5) # calculated and returned back to the call

print(area_cicle()) #Calling the function
````

````python
def abc():
  print("Hi")
  print("Hello")
  return  #function execution stops
  print("Hi") #unreachable as return was encountered before
  print("Hello")


abc()
````

### Parameters and Arguments

````python
#Parameter and Argument in Function

def area_circle(radius): #radius is a parameter
  print(3.14*radius*radius) #print happens here itself and then control goes back to the call

area_circle(8) #calling the function
area_circle(12) #12 is an argument
area_circle(4)
area_circle(10)
n=int(input())
area_circle(n)

````

### Scope of Variables

````python
#Scope of the variable

a=1 #global variable
def area_circle(radius):
  area=3.14*radius*radius #area -> local variable
  return area #once function ends, area variable is wiped out of the memory

b=area_circle(10) #global variable
print(a)
print(b)
print(area) #this gives error, as value was erased when function ended

````

````python
#Scope of the variable

a=1 #global variable - can be used anywhere in the program
def area_circle(radius):
  area=3.14*radius*radius #area -> local variable
  print(a)
  return area #once function ends, area variable is wiped out of the memory

b=area_circle(10) #global variable
print(a)
print(b)

````

````python
c=1 # Global variable
def abc():
    c=2 # new variable c with value 2
    print(c) #2, c variable is erased once the function ends

abc() # Calling the function
print(c) #Golbal variable with value 1 is printed
````

````python
c=1 # Global variable
def abc():
    global c #use global variable c value in the function
    c=2 # The global variable with updated value 2
    print(c) #2, c variable is erased once the function ends

abc() # Calling the function
print(c) #Golbal variable is reused, changes made inside the function are called
````

````python
rate =7
principal = int(input("How much do you want to invest? "))
time = float(input("How long do you want to invest it for? "))

def print_interest(principal, time):
    result = principal * time * rate
    return round(result,2)

print(print_interest(principal,time))
````

### Positional and keyword Arguments

````python
def bio_data(name,age,gender):
    print("Name: ", name)
    print("Age: ", age)
    print("Gender: ",gender)

bio_data(age= 23, gender="male",name="Kiran")
````

````python
#Keyword Arguments

def bio_data(name,age,gender): #23,"male","Balakrishna"
  print("Name: ", name)
  print("Age: ",age)
  print("Gender: ",gender)

bio_data(age=23,gender="male",name="Balakrishna")
````

````python
#Positional Arguments

def bio_data(name,age,gender): #23,"male","Balakrishna"
  print("Name: ", name)
  print("Age: ",age)
  print("Gender: ",gender)

bio_data("Balakrishna",23,"male")
````

````python
# both arguments can be passed together
# keyword arguments must always be before the positional arguments
bio_data("Vedkumar", age=23,"Male") #not allowed
````

````python
bio_data("Vedkumar", age=23,gender="Male")
````

````python
#Default and non default arguments

def bio_data(name,age,gender):
  print("Name: ", name)
  print("Age: ",age)
  print("Gender: ",gender)

bio_data("Aakar",23)
````
