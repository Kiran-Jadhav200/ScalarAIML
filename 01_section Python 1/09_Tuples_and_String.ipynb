# Tuples
A tuple is an ordered collection of elements that is immutable, meaning that once it is created, its contents cannot be changed. Tuples are defined by enclosing elements in parentheses `()` and separating them with commas.

```python
planets = ["Mercury", "Venus", "Earth", "Mars", "Jupiter", "Saturn", "Uranus", "Neptune"]
print(planets)
```

```python
planets.append("Sun")
print(planets)
```

```python
# a=()
a=(1,2,3,4,5)
print(type(a))
```

```python
d = ()  # empty tuple
print(type(d))
```

```python
b= (1) # not a tuple -> it's an integer in brackets
print(type(b))
e = (1,) # tuple with single element
print(type(e))
```

```python
# Creating tuples with the range keyword
a = list(range(10))
print(type(a))
aa = tuple(range(10))
print(type(aa))
```

```python
tu = (1, "2", 3.0, [4, 5], (6, 7)) # Tuples with heterogeneous data types
print(type(tu))
```

```python
# Immutability of tuples
print(aa)
print(aa[0]) # tuple_name[index] - access values using indices
print(aa[3])
```

```python
aa[1] = 20  # This will raise an error because tuples are immutable
```

```python
a=[[1,2],2]
print(a)
#   0 1 2 3  4
tu=(3,4,5,6,[1,2,3]) # tuple with list as an element
print(tu)

# if list is inside a tuple and lists are mutable
# I can mutate the list inside
print(tu[4])
tu[4].append('c') # this is possible
print(tu)
```

## ASCII - American Standard Code of Information Interchange

- a - z: 97 to 122
- A - Z: 65 to 90
- 0 - 9: 48 to 57 -> "0" != 0


```python
# ord -> gives ASCII value of string
print(ord("A"))
print(ord("0"))
print(ord("z"))
print(ord("*"))
```

```python
# chr -> gives character value of ASCII
print(chr(65))
print(chr(48))
print(chr(122))
print(chr(42))
```
